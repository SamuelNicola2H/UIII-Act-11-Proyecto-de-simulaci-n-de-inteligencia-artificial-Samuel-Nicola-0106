```python
# ============================================================
# UIII Act 11 Proyecto de simulación de inteligencia artificial
# Uso de diccionarios, if, else, elif, while, funciones y menú
#
# Tema: Bombero
# ============================================================

# Diccionario vacío
bomberos = {}

# Lista donde se guardarán los registros
lista_bomberos = []

# ------------------------------------------------------------
# FUNCIÓN PARA REGISTRAR BOMBERO
# ------------------------------------------------------------

def registrar_bombero():

    print("\n===================================")
    print(" REGISTRO DE BOMBERO ")
    print("===================================")

    nombre = input("Nombre del bombero: ")
    estacion = input("Estación: ")
    rango = input("Rango: ")
    equipo = input("Equipo de protección: ")
    incendios = input("Incendios atendidos: ")
    especialidad = input("Especialidad en rescate: ")
    turno = input("Turno: ")

    # Diccionario con llaves y valores
    bomberos = {

        "Nombre": nombre,
        "Estación": estacion,
        "Rango": rango,
        "Equipo de protección": equipo,
        "Incendios atendidos": incendios,
        "Especialidad en rescate": especialidad,
        "Turno": turno
    }

    # Guardar en la lista
    lista_bomberos.append(bomberos)

    print("\n✅ Bombero registrado correctamente")


# ------------------------------------------------------------
# FUNCIÓN PARA MOSTRAR BOMBEROS
# ------------------------------------------------------------

def mostrar_bomberos():

    print("\n======================================================================================================")
    print("                                      LISTA DE BOMBEROS")
    print("======================================================================================================")

    if len(lista_bomberos) == 0:

        print("⚠ No hay bomberos registrados")

    else:

        print(f"{'NOMBRE':15} {'ESTACIÓN':15} {'RANGO':15} {'INCENDIOS':15} {'ESPECIALIDAD':20} {'TURNO':10}")

        print("------------------------------------------------------------------------------------------------------")

        # Ciclo for
        for b in lista_bomberos:

            print(f"{b['Nombre']:15} "
                  f"{b['Estación']:15} "
                  f"{b['Rango']:15} "
                  f"{b['Incendios atendidos']:15} "
                  f"{b['Especialidad en rescate']:20} "
                  f"{b['Turno']:10}")


# ------------------------------------------------------------
# FUNCIÓN IA
# ------------------------------------------------------------

def analizar_bombero():

    print("\n===================================")
    print(" ANALISIS IA DE BOMBERO ")
    print("===================================")

    nombre = input("Ingresa el nombre del bombero: ")

    encontrado = False

    for b in lista_bomberos:

        if b["Nombre"].lower() == nombre.lower():

            encontrado = True

            incendios = int(b["Incendios atendidos"])
            especialidad = b["Especialidad en rescate"].lower()

            print("\n🤖 Analizando información...")

            # if, elif y else
            if incendios >= 50 and especialidad == "altura":

                print("🔥 IA: Bombero experto en rescates de alto riesgo")

            elif incendios >= 20 and especialidad == "vehicular":

                print("🚒 IA: Especialista en accidentes vehiculares")

            elif incendios >= 10:

                print("🧯 IA: Bombero apto para combate de incendios")

            elif especialidad == "acuático":

                print("🌊 IA: Recomendado para rescates acuáticos")

            else:

                print("📘 IA: Bombero en entrenamiento")

    if encontrado == False:

        print("❌ Bombero no encontrado")


# ------------------------------------------------------------
# DICCIONARIO DE FUNCIONES
# ------------------------------------------------------------

menu_funciones = {

    "1": registrar_bombero,
    "2": mostrar_bomberos,
    "3": analizar_bombero
}


# ------------------------------------------------------------
# MENÚ PRINCIPAL
# ------------------------------------------------------------

while True:

    print("\n================================================")
    print(" 🤖 SISTEMA INTELIGENTE DE BOMBERO 🤖 ")
    print("================================================")

    print("1. Registrar bombero")
    print("2. Mostrar bomberos")
    print("3. Analizar bombero")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion in menu_funciones:

        menu_funciones[opcion]()

    elif opcion == "4":

        print("\n👋 Gracias por usar el sistema")
        break

    else:

        print("\n❌ Opción incorrecta")
```
